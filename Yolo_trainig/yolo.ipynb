```python
import os
import pandas as pd
import json
import yaml
from tqdm import tqdm

# --- CONFIGURAÇÃO ---
# O script assume a seguinte estrutura de pastas:
# ./converter_dataset.py
# ./dataset/
#      info.json
#      train/
#          data/
#              <imagens>.jpg
#          labels/
#              detections.csv

# Caminhos de entrada (geralmente não precisam de alteração se a estrutura acima for seguida)
BASE_DIR = 'open-images-v7'
INFO_JSON_PATH = os.path.join(BASE_DIR, 'info.json')
IMAGES_DIR = os.path.join(BASE_DIR, 'train/data')
CSV_PATH = os.path.join(BASE_DIR, 'train/detections.csv')

# Caminhos de saída
OUTPUT_DIR = 'yolo_dataset' # Pasta onde o novo dataset será criado
LABELS_OUTPUT_DIR = os.path.join(OUTPUT_DIR, 'labels')
YAML_OUTPUT_PATH = os.path.join(OUTPUT_DIR, 'dataset.yaml')

# --- LÓGICA DO SCRIPT ---

def create_yolo_dataset():
    """
    Converte um dataset no formato Open Images (baixado com FiftyOne)
    para o formato YOLOv8.
    """
    print("Iniciando a conversão do dataset para o formato YOLO...")

    # 1. Criar diretórios de saída
    print(f"Criando diretório de saída em: '{OUTPUT_DIR}'")
    os.makedirs(LABELS_OUTPUT_DIR, exist_ok=True)

    # 2. Carregar informações de classes do info.json
    try:
        with open(INFO_JSON_PATH, 'r') as f:
            info_data = json.load(f)
        class_names = info_data.get('classes', [])
        if not class_names:
            print(f"AVISO: A chave 'classes' não foi encontrada em '{INFO_JSON_PATH}'. Verifique o arquivo.")
            return
        class_to_id = {name: i for i, name in enumerate(class_names)}
        print(f"Sucesso! {len(class_names)} classes encontradas.")
    except FileNotFoundError:
        print(f"ERRO: Arquivo 'info.json' não encontrado em '{INFO_JSON_PATH}'. Verifique o caminho.")
        return

    # 3. Carregar e filtrar o CSV de anotações
    try:
        df = pd.read_csv(CSV_PATH)
        print(f"Total de anotações no CSV: {len(df)}")
    except FileNotFoundError:
        print(f"ERRO: Arquivo 'detections.csv' não encontrado em '{CSV_PATH}'. Verifique o caminho.")
        return

    # Garantir que estamos processando apenas imagens que realmente existem
    print("Filtrando anotações para imagens existentes na pasta 'data/'...")
    image_filenames_in_dir = set(os.listdir(IMAGES_DIR))
    image_ids_in_dir = {os.path.splitext(f)[0] for f in image_filenames_in_dir}
    
    original_rows = len(df)
    df_filtered = df[df['ImageID'].isin(image_ids_in_dir)].copy()
    print(f"Anotações retidas após filtro: {len(df_filtered)} (de {original_rows})")

    # Mapear nome do arquivo da imagem para seu ID (sem extensão)
    filename_to_id_map = {os.path.splitext(f)[0]: f for f in image_filenames_in_dir}
    df_filtered['ImageFilename'] = df_filtered['ImageID'].map(filename_to_id_map)

    # 4. Processar anotações e criar arquivos .txt
    print("Gerando arquivos de anotação .txt para cada imagem...")
    grouped = df_filtered.groupby('ImageID')
    
    for image_id, group in tqdm(grouped, desc="Processando imagens"):
        txt_filename = f"{image_id}.txt"
        txt_filepath = os.path.join(LABELS_OUTPUT_DIR, txt_filename)
        
        yolo_annotations = []
        for _, row in group.iterrows():
            class_name = row['LabelName']
            if class_name not in class_to_id:
                continue

            class_id = class_to_id[class_name]
            
            # Coordenadas do OpenImages CSV já são normalizadas (0.0 a 1.0)
            x_min, x_max, y_min, y_max = row['XMin'], row['XMax'], row['YMin'], row['YMax']
            
            # Converter para o formato YOLO (center_x, center_y, width, height)
            x_center = (x_min + x_max) / 2.0
            y_center = (y_min + y_max) / 2.0
            width = x_max - x_min
            height = y_max - y_min
            
            yolo_annotations.append(f"{class_id} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}")
            
        if yolo_annotations:
            with open(txt_filepath, 'w') as f:
                f.write('\n'.join(yolo_annotations))

    print(f"Total de {len(grouped)} arquivos .txt criados em '{LABELS_OUTPUT_DIR}'.")

    # 5. Criar o arquivo dataset.yaml
    print(f"Criando o arquivo de configuração '{YAML_OUTPUT_PATH}'...")
    
    # O caminho para as imagens deve ser relativo ao local do arquivo .yaml
    train_images_path = os.path.relpath(IMAGES_DIR, start=OUTPUT_DIR)
    
    yaml_data = {
        'path': os.path.abspath(OUTPUT_DIR), # Ultralytics recomenda um path absoluto
        'train': os.path.relpath(IMAGES_DIR, start=OUTPUT_DIR), # path/to/images
        'val': os.path.relpath(IMAGES_DIR, start=OUTPUT_DIR),  # Usando o mesmo para validação
        'nc': len(class_names),
        'names': class_names
    }
    
    with open(YAML_OUTPUT_PATH, 'w') as f:
        yaml.dump(yaml_data, f, sort_keys=False, default_flow_style=False)

    
        
    print("\n✅ Conversão concluída com sucesso!")
    print("\nEstrutura gerada:")
    print(f"  - {OUTPUT_DIR}/")
    print(f"    - labels/       (Contém os arquivos .txt de anotação)")
    print(f"    - dataset.yaml  (Seu arquivo de configuração)")
    print("\nPróximos passos:")
    print(f"  Para treinar seu modelo, use o comando:")
    print(f"  yolo train data={YAML_OUTPUT_PATH} model=yolov8n.pt epochs=100 imgsz=640")


if __name__ == '__main__':
    create_yolo_dataset()

```

```python

import os
import shutil
import random

# Diretórios
train_path = 'datasetCriado/train'
val_path = 'datasetCriado/val'

train_images_path = os.path.join(train_path, 'images')
train_labels_path = os.path.join(train_path, 'labels')
val_images_path = os.path.join(val_path, 'images')
val_labels_path = os.path.join(val_path, 'labels')

# Cria pasta val se não existir
os.makedirs(val_images_path, exist_ok=True)
os.makedirs(val_labels_path, exist_ok=True)

# Lista imagens no train
train_images = [f for f in os.listdir(train_images_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

# Seleciona aleatoriamente 10% das imagens para validação
val_size = int(len(train_images) * 0.1)
val_images = random.sample(train_images, val_size)

# Copia imagens e seus respectivos labels para a pasta val
for image in val_images:
    image_src = os.path.join(train_images_path, image)
    label_name = os.path.splitext(image)[0] + '.txt'
    label_src = os.path.join(train_labels_path, label_name)

    image_dst = os.path.join(val_images_path, image)
    label_dst = os.path.join(val_labels_path, label_name)

    shutil.move(image_src, image_dst)
    if os.path.exists(label_src):
        shutil.move(label_src, label_dst)

# Conta arquivos de label vazios (sem bounding box)
count_empty_labels = 0
for label_file in os.listdir(train_labels_path):
    if not label_file.endswith('.txt'):
        continue
    label_path = os.path.join(train_labels_path, label_file)
    with open(label_path, 'r') as f:
        lines = f.readlines()
        if len(lines) == 0:
            count_empty_labels += 1

print(f'Total de imagens no train: {len(train_images)}')
print(f'Total de imagens para validação: {len(val_images)}')
print(f'Total de labels vazios (sem bounding box): {count_empty_labels}')
```
